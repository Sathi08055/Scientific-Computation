```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.patches as patches
```

```python
t=np.linspace(0,2*np.pi,1000)
xt=np.cos(t)
yt=np.sin(t)
number=1000000
np.random.seed(21)

fig,ax=plt.subplots(figsize=(5,5))
rectangle=patches.Rectangle((-1,-1),2,2,color="yellow")
ax.add_patch(rectangle)
ax.set_xlim(-1.1,1.1)
ax.set_ylim(-1.1,1.1)
ax.plot(xt,yt,color="black")
points=(np.random.random(size=(number,2))-0.5)*2
mask1=np.linalg.norm(points,axis=1)<=1
points_in=points[mask1]
ax.scatter(points_in[:,0],points_in[:,1],s=1,label=f"$N_{{in}}= {points_in.shape[0]}$\n$N_{{Total}}={points.shape[0]}$")
plt.legend()
plt.show()
print(points_in.shape[0]/points.shape[0]*4)
```
