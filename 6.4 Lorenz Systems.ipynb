```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.integrate as integrate
```

```python
a=10
b=60
c=3

def fODE(t,X,a,b,c):
    x,y,z=X
    x1 = a*(y-x)
    y1 = x*(b-z)-y
    z1 = x*y-c*z
    return [x1,y1,z1]

t=np.linspace(0,30,10000)
sol=integrate.solve_ivp(fODE,t_span=[t.min(),t.max()],t_eval=t,y0=[0,1,-2],args=[a,b,c])
x=sol.y[0]
y=sol.y[1]
z=sol.y[2]
fig=plt.figure()
ax=fig.add_subplot(projection="3d")
ax.plot3D(x,y,z)
plt.show()


x1 = a*(y-x)
y1 = x*(b-z)-y
z1 = x*y-c*z
r=np.column_stack((x1,y1,z1))

rmag=np.linalg.norm(r,axis=1)
print(rmag.shape)
fig=plt.figure()
ax=fig.add_subplot()
ax.plot(t,rmag)
plt.title("magnitute over time ")
plt.show()
```
