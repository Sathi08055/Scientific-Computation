```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
pA=1e-4
pB=0.1
name="Secular Equilibrium"
NA=5_000_000
NB=0
NC=0
nA=[]
nB=[]
nC=[]
steps=500
np.random.seed(10)
for i in range(steps):
    decayA=np.random.binomial(NA,pA)
    decayB=np.random.binomial(NB,pB)
    if(NA>=0):
        NA-= decayA
    if(NB>=0):
        NB+= decayA -decayB
        NC+= decayB
    nA.append(NA)
    nB.append(NB)
    nC.append(NC)
nA=np.array(nA)
nB=np.array(nB)

plt.figure(figsize=(10,4.5),dpi=200)
plt.plot(nA*pA,label=f"$P_A$={pA}")
plt.plot(nB*pB,label=f"$P_B$={pB}")
plt.legend(loc="upper right")
plt.yscale("log")
plt.title(f"{name}\n")
plt.ylabel("Activity")
plt.xlabel("time")
plt.show()
```

```python
pA1=0.1
pA2=0.01
pA3=1e-4
NA1=5_000_000
NA2=NA1
NA3=NA1
nA1=[]
nA2=[]
nA3=[]
steps=500
np.random.seed(10)
for i in range(steps):
    decayA=np.random.binomial(NA1,pA1)
    decayB=np.random.binomial(NA2,pA2)
    decayC=np.random.binomial(NA3,pA3)
    NA1-=decayA
    NA2-=decayB
    NA3-=decayC
    nA1.append(NA1)
    nA2.append(NA2)
    nA3.append(NA3)
nA=np.array(nA1)
nB=np.array(nA2)
nC=np.array(nA3)

plt.figure(figsize=(10,4.5),dpi=200)
plt.plot(nA*pA1,label=f"$P_A$={pA1}")
plt.plot(nB*pA2,label=f"$P_B$={pA2}")
plt.plot(nC*pA3,label=f"$P_B$={pA3}")
plt.legend(loc="upper right")
plt.yscale("log")
plt.title(f"Radioactivity\n")
plt.ylabel("Activity")
plt.xlabel("time")
plt.show()
```
